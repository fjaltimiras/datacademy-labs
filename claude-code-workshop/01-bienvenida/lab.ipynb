# Comprueba el reporte

Tres sucursales, tres archivos y boletas repetidas. Antes de confiar en un total, se comprueba.

```python
# Solo en Colab o Jupyter: baja los CSV de la unidad si no están en la carpeta.
import os, urllib.request

BASE = 'https://raw.githubusercontent.com/fjaltimiras/datacademy-labs/main/claude-code-workshop/01-bienvenida/'
for sucursal in ['centro', 'norte', 'sur']:
    archivo = f'ventas_{sucursal}.csv'
    if not os.path.exists(archivo):
        urllib.request.urlretrieve(BASE + archivo, archivo)
```

## 1. Mirar los datos

Abrimos los tres archivos y los juntamos en una sola tabla.

```python
import pandas as pd

partes = []
for sucursal in ['centro', 'norte', 'sur']:
    tabla = pd.read_csv(f'ventas_{sucursal}.csv')
    tabla['sucursal'] = sucursal.capitalize()
    partes.append(tabla)

todo = pd.concat(partes, ignore_index=True)
print('Filas en total:', len(todo))
todo.head()
```

## 2. Buscar boletas repetidas

```python
repetidas = todo[todo.duplicated(keep=False)].sort_values('boleta')
print('Boletas repetidas:', repetidas['boleta'].nunique())
repetidas
```

## 3. ¿Cuánto cambia el total?

```python
limpio = todo.drop_duplicates()
print('Con repetidas: $', todo['total'].sum())
print('Sin repetidas: $', limpio['total'].sum())
print('Filas:', len(todo), '->', len(limpio))
```

## 4. Total por sucursal

```python
import matplotlib.pyplot as plt

por_sucursal = limpio.groupby('sucursal')['total'].sum()
por_sucursal.plot.bar(color='#00a383', rot=0, title='Ventas de octubre (datos sintéticos)')
plt.ylabel('Pesos')
plt.show()
por_sucursal
```

## 5. Tu comprobación

Cambia `0` por el total de la sucursal **Sur** y ejecuta la celda.

```python
mi_total_sur = 0

if mi_total_sur == por_sucursal['Sur']:
    print('Correcto: coincide con los datos.')
else:
    print('Todavía no coincide. Revisa el resultado del paso 4.')
```
