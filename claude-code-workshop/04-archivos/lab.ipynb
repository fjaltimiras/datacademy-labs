# Limpia y cuadra

Mirar, acordar, aplicar y cuadrar, con clientes inventados.

```python
# Solo en Colab o Jupyter: baja los CSV si no están en la carpeta.
import os, urllib.request

BASE = 'https://raw.githubusercontent.com/fjaltimiras/datacademy-labs/main/claude-code-workshop/04-archivos/'
for archivo in ['clientes.csv']:
    if not os.path.exists(archivo):
        urllib.request.urlretrieve(BASE + archivo, archivo)
```

## 1. Mirar

```python
import pandas as pd

clientes = pd.read_csv('clientes.csv')
print('Filas:', len(clientes))
print('Códigos repetidos:', clientes['codigo'].duplicated().sum())
print('Comunas vacías:', clientes['comuna'].isna().sum())
clientes.head(8)
```

## 2. Aplicar las reglas

Una fila por código, textos sin espacios sobrantes y comuna vacía como PENDIENTE. Si un código repetido trae la comuna en otra fila, se conserva esa comuna.

```python
limpio = clientes.copy()
limpio['cliente'] = limpio['cliente'].str.split().str.join(' ').str.title()
limpio['comuna'] = limpio['comuna'].str.title()
limpio = limpio.groupby('codigo', as_index=False).first()
limpio['comuna'] = limpio['comuna'].fillna('PENDIENTE')
limpio.head(8)
```

## 3. Cuadrar

```python
salieron = len(clientes) - len(limpio)
print('Entraron', len(clientes), '· salieron', salieron, '· quedan', len(limpio))
print('Total de compras:', limpio['compras'].sum())
limpio[limpio['comuna'] == 'PENDIENTE']
```

## 4. Tu comprobación

¿Cuánto habría inflado el total no quitar las repetidas? Cambia `0` y ejecuta.

```python
mi_diferencia = 0

if mi_diferencia == clientes['compras'].sum() - limpio['compras'].sum():
    print('Correcto: coincide con los datos.')
else:
    print('Todavía no coincide. Suma la columna compras antes y después.')
```
