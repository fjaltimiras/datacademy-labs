## 1. Las reglas y los casos

```python
import pandas as pd

print(open('reglas.md', encoding='utf-8').read())
casos = pd.read_csv('casos.csv', dtype=str, keep_default_na=False)
casos
```

## 2. Resultados esperados

Calcula primero tres casos a mano. Después ejecuta.

```python
import math

def referencia(zona, kg, compra):
    try:
        kg, compra = float(kg), float(compra)
    except ValueError:
        return 'ERROR'
    if zona not in ('Santiago', 'Regiones') or kg <= 0 or compra < 0:
        return 'ERROR'
    if zona == 'Santiago' and compra >= 60000:
        return '0'
    base = 4000 if zona == 'Santiago' else 6500
    return str(base + max(0, math.ceil(kg - 5)) * 700)

esperado = {f.caso: referencia(f.zona, f.kg, f.compra) for f in casos.itertuples()}
print(len(esperado), 'casos listos')
```

## 3. Tu función

Esta versión tiene cuatro errores. Corrígela hasta que coincidan los 10 casos.

```python
def mi_cotizador(zona, kg, compra):
    kg = float(kg or 0)
    compra = float(compra)
    if compra > 60000:
        return '0'
    base = 4000 if zona == 'Santiago' else 6500
    extra = max(0, int(kg - 5))
    return str(base + extra * 700)

buenos = 0
for f in casos.itertuples():
    obtenido = mi_cotizador(f.zona, f.kg, f.compra)
    ok = obtenido == esperado[f.caso]
    buenos += ok
    if not ok:
        print(f.caso, '· obtenido', obtenido, '· esperado', esperado[f.caso])
print(buenos, 'de', len(casos), 'coinciden')
```
